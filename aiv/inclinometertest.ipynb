```python
import os
from pathlib import Path

import pandas as pd
from scipy import stats

# Supply the directory containing the external TMA*.xls test workbooks.
DATA_DIR = Path(os.environ.get("INCLINOMETER_DATA_DIR", "data/inclinometer"))
workbooks = [
    (270.0, "TMA270.xls"),
    (180.0, "TMA180.xls"),
    (90.0, "TMA90.xls"),
    (0.0, "TMA0.xls"),
    (-90.0, "TMA-90.xls"),
    (-180.0, "TMA-180.xls"),
    (-90.0, "TMA-90reverse.xls"),
    (0.0, "TMA0reverse.xls"),
    (90.0, "TMA90reverse.xls"),
    (180.0, "TMA180reverse.xls"),
    (207.0, "TMA207reverse.xls"),
]
azencoder = []
x = []
y = []
azavg = []
xavg = []
yavg = []
xrms = []
yrms = []

for az, filename in workbooks:
    dataframe = pd.read_excel(DATA_DIR / filename)
    xmeas = pd.to_numeric(dataframe["X-Axis Value"].astype(str).str.replace('"', "", regex=False))
    ymeas = pd.to_numeric(dataframe["Y-Axis Value"].astype(str).str.replace('"', "", regex=False))

    azencoder.extend([az] * len(xmeas))
    x.extend(xmeas.astype(float).tolist())
    y.extend(ymeas.astype(float).tolist())
    azavg.append(az)
    xavg.append(stats.mean(xmeas))
    yavg.append(stats.mean(ymeas))
    xrms.append(stats.stdev(xmeas))
    yrms.append(stats.stdev(ymeas))
```

```python
print(yrms)
```

```python
plt.scatter(azavg, xavg, label="Average X measurements vs. Azimuth")
plt.errorbar(azavg, xavg, yerr=xrms, fmt="o")
plt.rcParams["figure.figsize"] = [14, 8]
# plt.gcf().autofmt_xdate()
# myFmt = mdates.DateFormatter('%H:%M')
# plt.gca().xaxis.set_major_formatter(myFmt)
plt.xlabel("Az encoder")
plt.ylabel("X tilt")
plt.title("inclinometer measurments")
plt.legend()
plt.show()
```

```python
plt.scatter(azavg, yavg, label="Average Y measurements vs. Azimuth")
plt.errorbar(azavg, yavg, yerr=yrms, fmt="o")
plt.rcParams["figure.figsize"] = [14, 8]
# plt.gcf().autofmt_xdate()
# myFmt = mdates.DateFormatter('%H:%M')
# plt.gca().xaxis.set_major_formatter(myFmt)
plt.xlabel("Az encoder")
plt.ylabel("Y tilt")
plt.title("inclinometer measurments")
plt.legend()
plt.show()
```
